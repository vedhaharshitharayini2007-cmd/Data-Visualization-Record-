```python
from PIL import Image as PILImage

image_path = 'cat im.jpeg (1).jpeg'
original_img = PILImage.open(image_path)

# Further adjusted coordinates for only the eyes, no forehead, no nose, both eyes fully visible
left = 160
upper = 180
right = 500
lower = 260

eyes_region = (left, upper, right, lower)
cropped_img = original_img.crop(eyes_region)

display(cropped_img)
```

```python
flipped_h = original_img.transpose(PILImage.FLIP_LEFT_RIGHT)
display(flipped_h)
```

### Horizontally Flipped Image

```python
flipped_v = original_img.transpose(PILImage.FLIP_TOP_BOTTOM)
display(flipped_v)
```

```python
import numpy as np
from google.colab import files

# Upload CSV file
uploaded = files.upload()

# Read CSV (numeric data only)
filename = list(uploaded.keys())[0]
data = np.genfromtxt(filename, delimiter=',', skip_header=1)

print("CSV Data:")
print(data)

# Statistical Operations
print("\nMean of each column:")
print(np.mean(data, axis=0))

print("\nMedian of each column:")
print(np.median(data, axis=0))

print("\nMaximum of each column:")
print(np.max(data, axis=0))

print("\nMinimum of each column:")
print(np.min(data, axis=0))

# Comparison Operation
print("\nRows where first column > its mean:")
print(data[data[:,0] > np.mean(data[:,0])])
```
