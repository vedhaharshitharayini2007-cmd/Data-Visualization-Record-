```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

uploaded = files.upload()

df = pd.read_csv('house_prices.csv')

print(df.head())
print(df.columns)
```

```python
df['price'].head(20).plot(kind='line')
plt.title('House Prices')
plt.show()
```

```python
df.groupby('bedrooms')['price'].mean().plot(kind='bar')
plt.title('Price by Bedrooms')
plt.show()
```

```python
plt.hist(df['price'], bins=20)
plt.title('Price Distribution')
plt.show()
```

```python
sns.kdeplot(df['price'])
plt.title('Price Density')
plt.show()
```

```python
plt.scatter(df['sqft_living'], df['price'])
plt.xlabel('Living Area')
plt.ylabel('Price')
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('house_prices.csv')

sns.scatterplot(
    data=df,
    x='sqft_living',
    y='price',
    hue='bedrooms',
    size='bathrooms',
    style='waterfront',
    alpha=0.6
)

plt.title('House Price Visualization')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('house_prices.csv')

df.groupby('bedrooms')['price'].mean().plot(kind='bar')
plt.show()

df.groupby('bathrooms')['price'].mean().plot(kind='bar')
plt.show()

df.groupby('floors')['price'].mean().plot(kind='bar')
plt.show()
```
